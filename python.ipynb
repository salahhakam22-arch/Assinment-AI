```python
def filter_longer(words, k):
    result = []
    for word in words:
        if len(word) > k:
            result.append(word)
    return result
 
print(filter_longer(["apple", "pie", "banana"], 3))
```

```python
def area(shape, a, b=0):
    if shape == "circle":
        return 3.14 * a * a
    elif shape == "rectangle":
        return a * b
    else:
        return "Unknown shape"
 
print("Circle Area:", area("circle", 5))
print("Rectangle Area:", area("rectangle", 4, 6))
```

```python
class Restaurant:
    def __init__(self):
        self.menu_items = {}
        self.book_table = {}
        self.customer_orders = {}
 
    def add_item_to_menu(self, name, price):
        self.menu_items[name] = price
 
    def book_tables(self, table_number, customer_name):
        if table_number in self.book_table:
            print("Table", table_number, "is already booked")
        else:
            self.book_table[table_number] = customer_name
 
    def customer_order(self, customer_name, item):
        if item not in self.menu_items:
            print(item, "is not on the menu")
        elif customer_name in self.customer_orders:
            self.customer_orders[customer_name].append(item)
        else:
            self.customer_orders[customer_name] = [item]
 
    def print_menu(self):
        print("--- Menu ---")
        for item in self.menu_items:
            print(item, ":", self.menu_items[item])
 
    def print_reservations(self):
        print("--- Table Reservations ---")
        for table in self.book_table:
            print("Table", table, ":", self.book_table[table])
 
    def print_orders(self):
        print("--- Customer Orders ---")
        for customer in self.customer_orders:
            print(customer, ":", self.customer_orders[customer])
 
 
r = Restaurant()
r.add_item_to_menu("Pizza", 80)
r.add_item_to_menu("Burger", 60)
r.add_item_to_menu("Juice", 20)
r.book_tables(1, "Ali")
r.book_tables(2, "Sara")
r.customer_order("Ali", "Pizza")
r.customer_order("Ali", "Juice")
r.customer_order("Sara", "Burger")
r.print_menu()
r.print_reservations()
r.print_orders()
```

```python
class Customer:
    def __init__(self, name, age):
        self.name = name
        self.age = age
 
    def getName(self):
        return self.name
 
    def setName(self, name):
        self.name = name
 
    def getAge(self):
        return self.age
 
    def setAge(self, age):
        self.age = age
 
 
class Bike:
    def __init__(self, name, price):
        self.name = name
        self.price = price
 
    def getName(self):
        return self.name
 
    def setName(self, name):
        self.name = name
 
    def getPrice(self):
        return self.price
 
    def setPrice(self, price):
        self.price = price
 
 
class MainBikeRental:
    def __init__(self):
        self.bikes = []
        self.rentals = []
 
    def getBikes(self):
        return self.bikes
 
    def setBikes(self, bikes):
        self.bikes = bikes
 
    def getRentals(self):
        return self.rentals
 
    def setRentals(self, rentals):
        self.rentals = rentals
 
    def addBike(self, bike):
        self.bikes.append(bike)
 
    def requestBike(self, customer, number_of_bikes):
        if customer.getAge() <= 6:
            print("Sorry, the customer must be older than 6 years")
        elif number_of_bikes <= len(self.bikes):
            for i in range(number_of_bikes):
                bike = self.bikes.pop()
                self.rentals.append(bike)
            print(customer.getName(), "rented", number_of_bikes, "bike(s)")
        else:
            print("Please wait 10 mins")
 
    def returnBike(self, rentalTime, num_of_bikes):
        if num_of_bikes > len(self.rentals):
            print("Error: you cannot return more bikes than were rented")
        else:
            bill = rentalTime * 40
            for i in range(num_of_bikes):
                bike = self.rentals.pop()
                self.bikes.append(bike)
            print("Total bill:", bill)
 
    def totalCost(self):
        total = 0
        for bike in self.bikes:
            total = total + bike.getPrice()
        return total
 
 
shop = MainBikeRental()
shop.addBike(Bike("BMX", 200))
shop.addBike(Bike("Mountain", 300))
shop.addBike(Bike("Road", 250))
print("Bikes in store:", len(shop.getBikes()))
print("Total cost of bikes in store:", shop.totalCost())
 
ali = Customer("Ali", 20)
kid = Customer("Sam", 5)
shop.requestBike(ali, 2)
print("Bikes in store:", len(shop.getBikes()))
shop.requestBike(ali, 5)
shop.requestBike(kid, 1)
shop.returnBike(3, 2)
print("Bikes in store:", len(shop.getBikes()))
print("Total cost of bikes in store:", shop.totalCost())
```
